```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
customers = pd.read_csv('../data/olist_customers_dataset.csv')
orders = pd.read_csv('../data/olist_orders_dataset.csv')
order_items = pd.read_csv('../data/olist_order_items_dataset.csv')
order_payments = pd.read_csv('../data/olist_order_payments_dataset.csv')
products = pd.read_csv('../data/olist_products_dataset.csv')

print("All datasets loaded!")
print(f"Customers: {customers.shape}")
print(f"Orders: {orders.shape}")
print(f"Order Items: {order_items.shape}")
print(f"Order Payments: {order_payments.shape}")
print(f"Products: {products.shape}")
```

```python
print("=== CUSTOMERS ===")
display(customers.head(3))

print("=== ORDERS ===")
display(orders.head(3))

print("=== ORDER ITEMS ===")
display(order_items.head(3))

print("=== ORDER PAYMENTS ===")
display(order_payments.head(3))

print("=== PRODUCTS ===")
display(products.head(3))
```

```python
print("=== MISSING VALUES ===\n")

datasets = {
    'Customers': customers,
    'Orders': orders,
    'Order Items': order_items,
    'Order Payments': order_payments,
    'Products': products
}

for name, df in datasets.items():
    missing = df.isnull().sum().sum()
    print(f"{name}: {missing} missing values")
    if missing > 0:
        print(df.isnull().sum()[df.isnull().sum() > 0])
    print()
```

```python
# Fill missing product category with 'Unknown'
products['product_category_name'] = products['product_category_name'].fillna('Unknown')

# Fill missing product dimensions with median values
cols_to_fill = ['product_name_lenght', 'product_description_lenght', 
                'product_photos_qty', 'product_weight_g',
                'product_length_cm', 'product_height_cm', 'product_width_cm']

for col in cols_to_fill:
    products[col] = products[col].fillna(products[col].median())

# Convert date columns in orders to datetime format
date_cols = ['order_purchase_timestamp', 'order_approved_at',
             'order_delivered_carrier_date', 'order_delivered_customer_date',
             'order_estimated_delivery_date']

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col])

print("Data cleaned successfully!")
print(f"Products missing values remaining: {products.isnull().sum().sum()}")
print(f"Orders date columns type: {orders['order_purchase_timestamp'].dtype}")
```

```python
# Step 1: Merge orders with customers
df = orders.merge(customers, on='customer_id', how='left')

# Step 2: Merge with order items
df = df.merge(order_items, on='order_id', how='left')

# Step 3: Merge with payments
df = df.merge(order_payments, on='order_id', how='left')

# Step 4: Merge with products
df = df.merge(products, on='product_id', how='left')

print(f"Final merged dataframe shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
```

```python
# Filter only delivered orders
delivered = df[df['order_status'] == 'delivered']

print("=== BUSINESS OVERVIEW ===\n")
print(f"Total Orders:       {delivered['order_id'].nunique():,}")
print(f"Total Customers:    {delivered['customer_unique_id'].nunique():,}")
print(f"Total Revenue:      R$ {delivered['payment_value'].sum():,.2f}")
print(f"Avg Order Value:    R$ {delivered['payment_value'].mean():,.2f}")
print(f"Total Products:     {delivered['product_id'].nunique():,}")
print(f"Date Range:         {delivered['order_purchase_timestamp'].min().date()} to {delivered['order_purchase_timestamp'].max().date()}")
```

```python
# Extract year-month from purchase date
delivered['year_month'] = delivered['order_purchase_timestamp'].dt.to_period('M')

# Group by month and sum revenue
monthly_revenue = delivered.groupby('year_month')['payment_value'].sum().reset_index()
monthly_revenue['year_month'] = monthly_revenue['year_month'].astype(str)

# Plot
plt.figure(figsize=(14, 5))
plt.plot(monthly_revenue['year_month'], monthly_revenue['payment_value'], 
         marker='o', color='steelblue', linewidth=2)
plt.xticks(rotation=45, ha='right')
plt.title('Monthly Revenue Trend (2016-2018)', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Revenue (R$)')
plt.tight_layout()
plt.savefig('../visuals/monthly_revenue.png')
plt.show()
print("Chart saved!")
```

```python
# Group by category and sum revenue
top_categories = delivered.groupby('product_category_name')['payment_value'].sum()
top_categories = top_categories.sort_values(ascending=False).head(10).reset_index()
top_categories.columns = ['Category', 'Revenue']

# Plot
plt.figure(figsize=(12, 6))
sns.barplot(data=top_categories, x='Revenue', y='Category', palette='Blues_r')
plt.title('Top 10 Product Categories by Revenue', fontsize=14)
plt.xlabel('Total Revenue (R$)')
plt.ylabel('Category')
plt.tight_layout()
plt.savefig('../visuals/top_categories.png')
plt.show()
print("Chart saved!")
```

```python
# Group by payment type
payment_dist = delivered.groupby('payment_type')['order_id'].nunique().reset_index()
payment_dist.columns = ['Payment Type', 'Orders']
payment_dist = payment_dist.sort_values('Orders', ascending=False)

# Plot
plt.figure(figsize=(8, 6))
plt.pie(payment_dist['Orders'], 
        labels=payment_dist['Payment Type'], 
        autopct='%1.1f%%',
        colors=['steelblue', 'lightblue', 'dodgerblue', 'navy'],
        startangle=140)
plt.title('Payment Method Distribution', fontsize=14)
plt.tight_layout()
plt.savefig('../visuals/payment_distribution.png')
plt.show()
print("Chart saved!")
```

```python
# Group by state
top_states = delivered.groupby('customer_state')['customer_unique_id'].nunique()
top_states = top_states.sort_values(ascending=False).head(10).reset_index()
top_states.columns = ['State', 'Customers']

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=top_states, x='State', y='Customers', 
            hue='State', palette='Blues_r', legend=False)
plt.title('Top 10 States by Number of Customers', fontsize=14)
plt.xlabel('State')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.savefig('../visuals/customers_by_state.png')
plt.show()
print("Chart saved!")
```

```python
# Calculate delivery time in days
delivered['delivery_days'] = (
    delivered['order_delivered_customer_date'] - 
    delivered['order_purchase_timestamp']
).dt.days

# Group by state
delivery_by_state = delivered.groupby('customer_state')['delivery_days'].mean()
delivery_by_state = delivery_by_state.sort_values().head(10).reset_index()
delivery_by_state.columns = ['State', 'Avg Delivery Days']

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=delivery_by_state, x='State', y='Avg Delivery Days',
            hue='State', palette='Blues_r', legend=False)
plt.title('Top 10 States with Fastest Delivery', fontsize=14)
plt.xlabel('State')
plt.ylabel('Average Delivery Days')
plt.tight_layout()
plt.savefig('../visuals/delivery_by_state.png')
plt.show()
print("Chart saved!")
```

```python
# Group by order status
status_counts = df.groupby('order_status')['order_id'].nunique().reset_index()
status_counts.columns = ['Status', 'Count']
status_counts = status_counts.sort_values('Count', ascending=False)

# Plot
plt.figure(figsize=(10, 5))
sns.barplot(data=status_counts, x='Status', y='Count',
            hue='Status', palette='Blues_r', legend=False)
plt.title('Order Status Breakdown', fontsize=14)
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.savefig('../visuals/order_status.png')
plt.show()
print("Chart saved!")
```

```python
# Count orders per unique customer
customer_order_count = delivered.groupby('customer_unique_id')['order_id'].nunique()

# Classify
repeat = (customer_order_count > 1).sum()
one_time = (customer_order_count == 1).sum()

# Plot
plt.figure(figsize=(7, 7))
plt.pie([one_time, repeat],
        labels=['One-time Customers', 'Repeat Customers'],
        autopct='%1.1f%%',
        colors=['steelblue', 'lightblue'],
        startangle=90)
plt.title('Repeat vs One-Time Customers', fontsize=14)
plt.tight_layout()
plt.savefig('../visuals/repeat_customers.png')
plt.show()
print("Chart saved!")
```
